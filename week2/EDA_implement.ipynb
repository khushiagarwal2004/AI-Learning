```python
dataset.loc[:122,'Region']=0
dataset.loc[122:,'Region']=1
df=dataset
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
# Remove the null values
df=df.dropna().reset_index(drop=True)
```

```python
# Remove the 122 row
df=df.drop(122).reset_index(drop=True)
```

```python
# Fix spaces in column names
df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
# Change the required columns as integer datatype
df[['day','year','month','Temperature','RH','Ws']]=df[['day','year','month','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
df.columns
```

```python
# Changing the other columns to floor datatype
objects=[features for features in df.columns if df[features].dtypes=='string']
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```
