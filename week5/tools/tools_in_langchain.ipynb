```python
# Step 1 - create a function
def multiply(a,b):
    """Multiplies two numbers a and b"""
    return a*b
```

```python
# Step 2 - add type hints
def multiply(a:int,b:int)->int:
    """Multiplies two numbers a and b"""
    return a*b
```

```python
# Step 3 - add the tool decorator
@tool
def multiply(a:int,b:int)->int:
    """Multiplies two numbers a and b"""
    return a*b
```

```python
# Using Structured Tools with LLMs
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True,description="The first number to add")
    b: int = Field(required=True,description="The second number to add")
```

```python
def multiply_func(a:int,b:int)->int:
    return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiplies two numbers a and b",
    args_schema=MultiplyInput
)
```

```python
result=multiply_tool.invoke({'a':2,'b':3})
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```

```python
# Using BaseTool Class
from langchain_core.tools import BaseTool
from pydantic import BaseModel, Field
from typing import Type
```

```python
# args schema using pydantic model
class MultiplyInput(BaseModel):
    a: int = Field(required=True,description="The first number to add")
    b: int = Field(required=True,description="The second number to add")
    
```

```python
class MultiplyTool(BaseTool):
    name:str='multiply'
    description:str="Multiplies two numbers a and b"
    args_schema:Type[BaseModel]=MultiplyInput

    def _run(self,a:int,b:int)->int:
        return a*b
   
```

```python
multiply_tool=MultiplyTool()
```

```python
result=multiply_tool.invoke({'a':2,'b':3})
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

```python
# Toolkit
from langchain_core.tools import tool
# custom tool
@tool
def add(a:int,b:int)->int:
    """Adds two numbers a and b"""
    return a+b

@tool
def multiply(a:int,b:int)->int:
    """Multiplies two numbers a and b"""
    return a*b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add,multiply]
```

```python
toolkit=MathToolkit()
tools=toolkit.get_tools()
for tool in tools:
    print(tool.name,'=>',tool.description)
```
